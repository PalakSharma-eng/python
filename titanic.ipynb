```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv("titanic-dataset.csv")
data.describe(include="all")
```

```python
data[data["Sex"] == "male"]
```

```python
data[(data["Age"] > 30) & (data["Sex"] == "female") & (data["Pclass"] == 3)]
```

```python
data[data["Age"] > 30 | (data["Fare"] > 100)]
```

```python
data.drop(["Cabin", "Ticket"], axis=1, inplace=True)
```

```python
data.rename(columns={"Pclass": "Passenger Class", "SibSp": "Siblings/Spouses Aboard", "Parch": "Parents/Children Aboard"}, inplace=True)        
```

```python
data.head()
```

```python
data.rename(columns={"Passenger Class": "Pclass", "Siblings/Spouses Aboard": "SibSp", "Parents/Children Aboard": "Parch"}, inplace=True)
```

```python
data.head()
```

```python
data["Age"].fillna(data["Age"].mean(), inplace=True)    
```

```python
data["FamilySize"] = data["SibSp"] + data["Parch"] + 1
```

```python
pd.pivot_table(data, values="Survived", index="Sex", columns="Pclass", aggfunc="mean")
```

```python
data["Sex"] = data["Sex"].replace({"male": "M", "female": "F"})
```

```python
data["Sex"] = data["Sex"].replace({"male": "1", "female": "0"})
```

```python
data['Sex'].value_counts()
```

```python
import numpy as np

data["SurvivalStatus"] = np.where(data["Survived"] == 1, "Survived", "Not Survived")
```

```python
data['SurvivalStatus'].value_counts()
```

```python
df2 = data.copy()
```

```python
df1 = data[["PassengerId", "Name"]]
df2 = data[["PassengerId", "Survived"]]

result = pd.merge(df1, df2, on="PassengerId", how="inner")

pd.merge(df1, df2, how="inner")
pd.merge(df1, df2, how="outer")
pd.merge(df1, df2, how="left")
pd.merge(df1, df2, how="right")
```

```python
data.to_csv("titanic-dataset-cleaned.csv", index=False)
```
