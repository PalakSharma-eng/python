```python
import numpy as np
import pandas as pd
from scipy import stats

np.random.seed(42)
n = 40
df = pd.DataFrame({
    "Student_ID": range(1, n + 1),
    "Branch": np.random.choice(["CSE", "ECE", "ME", "CE"], n),
    "Study_Hours": np.round(np.random.uniform(1, 8, n), 1),
    "Attendance": np.random.randint(55, 100, n),
})
df["Marks"] = np.clip(
    np.round(30 + 5 * df["Study_Hours"] + 0.2 * df["Attendance"] + np.random.normal(0, 6, n)),
    0, 100
).astype(int)
df.head()
```

## Task 1

```python
def get_grade(marks):
    if marks >= 75:
        return "A"
    if marks >= 60:
        return "B"
    if marks >= 40:
        return "C"
    return "F"

df["Grade"] = df["Marks"].apply(get_grade)
print(df["Grade"].value_counts())
```

## Task 2

```python
low_attendance_marks = [marks for marks, attendance in zip(df["Marks"], df["Attendance"]) if attendance < 75]
marks_above_70 = list(filter(lambda marks: marks > 70, df["Marks"]))
graced_marks = list(map(lambda marks: min(marks + 5, 100), df["Marks"]))
print(low_attendance_marks)
print(marks_above_70)
print(graced_marks)
```

## Task 3

```python
def summary(*values, round_to=2):
    if not values:
        return {"count": 0, "minimum": None, "maximum": None, "average": None}
    return {
        "count": len(values),
        "minimum": min(values),
        "maximum": max(values),
        "average": round(sum(values) / len(values), round_to)
    }

print(summary(10, 20, 30))
print(summary(*df["Marks"]))
```

## Task 4

```python
df.to_csv("students.csv", index=False)

def load_data(filename):
    try:
        return pd.read_csv(filename)
    except FileNotFoundError:
        print(f"File '{filename}' was not found.")
        return None

print(load_data("students.csv").head())
print(load_data("missing.csv"))
```

## Task 5

```python
branch_averages = df.groupby("Branch")[["Marks", "Study_Hours"]].mean()
print(branch_averages)
print("Highest average marks:", branch_averages["Marks"].idxmax())
print(df.sort_values("Marks", ascending=False).head(5))
```

## Task 6

```python
marks = df["Marks"].to_numpy()
z_scores = (marks - marks.mean()) / marks.std()
print(z_scores)
print("Z-scores above 1:", np.sum(z_scores > 1))
```

## Task 7

```python
q1 = df["Marks"].quantile(0.25)
q3 = df["Marks"].quantile(0.75)
iqr = q3 - q1
lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr
outliers = df[(df["Marks"] < lower_limit) | (df["Marks"] > upper_limit)]
print("Mean:", df["Marks"].mean())
print("Median:", df["Marks"].median())
print("Mode:", df["Marks"].mode().tolist())
print("Variance:", df["Marks"].var())
print("Standard deviation:", df["Marks"].std())
print("Minimum:", df["Marks"].min())
print("Maximum:", df["Marks"].max())
print("Q1:", q1, "Q3:", q3, "IQR:", iqr)
print("Lower limit:", lower_limit, "Upper limit:", upper_limit)
print("Outliers:", outliers["Marks"].tolist())
```

The mean is the class average, while the median is the middle score after sorting.
The standard deviation measures how widely students’ scores vary around the mean.

## Task 8

```python
study_corr = df["Marks"].corr(df["Study_Hours"])
attendance_corr = df["Marks"].corr(df["Attendance"])
print("Study hours correlation:", study_corr)
print("Attendance correlation:", attendance_corr)
print("Stronger relationship:", "Study_Hours" if abs(study_corr) > abs(attendance_corr) else "Attendance")
```

The factor with the larger absolute correlation has the stronger linear relationship with marks. A positive correlation means both tend to increase together; a negative correlation means one tends to decrease as the other increases. Correlation alone does not prove causation because other factors may influence both variables.

## Task 9

```python
mean_marks = df["Marks"].mean()
sem = stats.sem(df["Marks"])
ci = stats.t.interval(0.95, df=len(df) - 1, loc=mean_marks, scale=sem)
print("95% confidence interval:", ci)
```

The 95% confidence interval gives a range of plausible values for the population mean marks, using a method that captures the true mean in 95% of repeated samples.

## Task 10

H0: The population mean marks equal 60.  \nH1: The population mean marks do not equal 60.

```python
t_stat, p_value = stats.ttest_1samp(df["Marks"], 60)
print("t-statistic:", t_stat)
print("p-value:", p_value)
print("Conclusion:", "Reject H0; the mean differs significantly from 60." if p_value < 0.05 else "Fail to reject H0; insufficient evidence that the mean differs from 60.")
```

## Task 11

H0: CSE and ECE students have equal population mean marks.  \nH1: CSE and ECE students have different population mean marks.

```python
cse_marks = df.loc[df["Branch"] == "CSE", "Marks"]
ece_marks = df.loc[df["Branch"] == "ECE", "Marks"]
t_stat_2, p_value_2 = stats.ttest_ind(cse_marks, ece_marks)
print("CSE mean:", cse_marks.mean(), "ECE mean:", ece_marks.mean())
print("t-statistic:", t_stat_2)
print("p-value:", p_value_2)
print("Conclusion:", "Reject H0; the groups differ significantly." if p_value_2 < 0.05 else "Fail to reject H0; no statistically significant difference detected.")
```
