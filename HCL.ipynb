```python
import numpy as np
import pandas as pd
import scipy.stats as stats


np.random.seed(10)
n = 30
canteen = pd.DataFrame({
    "Weekday": np.random.choice(["Mon", "Tue", "Wed", "Thu", "Fri"], n),
    "Meal": np.random.choice(["Breakfast", "Lunch", "Snacks"], n),
    "Orders": np.random.randint(40, 200, n),
})
price = canteen["Meal"].map({"Breakfast": 40, "Lunch": 90, "Snacks": 30})
canteen["Revenue"] = (canteen["Orders"] * price / 1000).round(2)    # in Rs. thousands
canteen["Queue"] = np.clip(
    np.round(2 + 0.08 * canteen["Orders"] + np.random.normal(0, 2, n)), 0, None
).astype(int)                                                       # people waiting
canteen.loc[7, "Queue"] = 45        # a special event: one unusually long queue

print(canteen.head())
print("Shape:", canteen.shape)
```
