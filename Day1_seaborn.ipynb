```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names()
iris = sns.load_dataset('iris')
titanic = sns.load_dataset('titanic')
penguins = sns.load_dataset('penguins')
flights = sns.load_dataset('flights')
tips = sns.load_dataset('tips')
```

```python
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex')
plt.show()
```

```python
titanic.head(1)
```

```python
sns.scatterplot(data=flights, x='year', y='passengers')
plt.show()
```

```python
sns.lineplot(data=titanic, x='age', y='fare', hue='class')
plt.show()
```

```python
sns.barplot(data=titanic, x='survived', y='fare', hue='sex')
plt.show()
```

```python
iris['species'].value_counts()
```

```python
sns.countplot(data=iris, x='species')
plt.show()
```

```python
sns.countplot(data=tips, x='smoker', hue='time')
plt.show()
```

```python
sns.histplot(data=tips, x='total_bill')
plt.show()
```

```python
sns.histplot(data=titanic, x='age', bins=10, kde=True)
plt.show()
```

```python
sns.kdeplot(data=titanic, x='age')
plt.show()
```
